```python
from transformers import AutoTokenizer

qwen = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-1.5B")
gpt2 = AutoTokenizer.from_pretrained("openai-community/gpt2")

print(f"Qwen2.5 kennt {len(qwen):_} Token, GPT-2 {len(gpt2):_}".replace("_", "'"))
```

Wie auf der Folie ein Codec: `encode` macht aus Text ein Array von Ganzzahlen, `decode` macht daraus wieder Text.

```python
ids = qwen.encode("Der Build ist fehlgeschlagen, weil ein Test rot ist.")
print(ids)
print(qwen.decode(ids))
```

## Die Beispiele der Folie

`zeige` zerlegt einen Text und zeigt die Token abwechselnd hinterlegt, Leerzeichen als ·, darunter die IDs. Den deutschen Satz und die Kantonalbank kennt ihr von der Folie, die anderen sind neu.

## Eine eigene Java-Zeile

> **Aufgabe 2**: Ersetzt die Zeile in der Zelle durch eine aus eurem Code, gern mit Einrückung, Annotation und langen Bezeichnern. Was ist ein einziges Token, was wird zerlegt? Welcher der beiden Tokenizer braucht für die Einrückung mehr Token?

```python
zeile = "        @Autowired private KundenRepository kundenRepository;"
zeige(zeile, qwen, gpt2)
```

```python
# @title Lösung (aufklappen)
# Häufige Schlüsselwörter wie private, static, final oder int sind je ein Token, mit dem
# Leerzeichen davor. Eigene Bezeichner werden zerlegt, oft an camelCase und Unterstrichen.
# Derselbe Wortteil hat je nach Leerzeichen und Grossschreibung eine andere ID:
# " Kunden" ist bei Qwen2.5 ein Token, in "kundenRepository" wird daraus k|unden.
# Einrückung: Qwen2.5 fasst die Leerzeichen zu einem Token zusammen, GPT-2 braucht für
# jedes ein eigenes. Tief eingerückter Code kostet bei GPT-2 deutlich mehr Token:
beispiel_java = "                return kundenRepository.findById(kundenId).orElseThrow();"
zeige(beispiel_java, qwen, gpt2)
```

## Derselbe Satz auf Deutsch und Englisch

> **Aufgabe 3**: Schreibt einen Satz auf Deutsch und seine englische Übersetzung. Vorher schätzen: Wie viel mehr Token braucht Deutsch? Dann ausführen und den Faktor bei Qwen2.5 und GPT-2 ablesen. Was heisst das für Kontextfenster und Preise?

```python
deutsch = "Der Kunde hat seine Adresse geändert."
englisch = "The customer has changed their address."
zeige(deutsch, qwen, gpt2)
zeige(englisch, qwen, gpt2)
for name, tokenizer in [("Qwen2.5", qwen), ("GPT-2", gpt2)]:
    faktor = len(tokenizer.encode(deutsch)) / len(tokenizer.encode(englisch))
    print(f"{name}: Deutsch braucht {faktor:.1f}-mal so viele Token wie Englisch")
```

```python
# @title Lösung (aufklappen)
# Deutsch braucht mehr Token, weil Tokenizer mit viel englischem Text gebaut werden und
# deutsche Wörter deshalb öfter zerlegt werden (Petrov et al. 2023). Bei GPT-2 ist der
# Unterschied grösser: Seine Trainingsdaten waren fast nur englisch (Radford et al. 2019).
# Kontextfenster, Limits und Preise zählen in Token. Derselbe Inhalt auf Deutsch füllt
# das Fenster schneller und kostet mehr. Die Faustregel von vier Zeichen pro Token gilt
# für Englisch.
paare = [
    ("Der Build ist fehlgeschlagen, weil ein Test rot ist.", "The build failed because a test is red."),
    ("Die Pipeline ist grün, aber der Deploy auf Produktion ist fehlgeschlagen.",
     "The pipeline is green, but the deployment to production failed."),
]
for de, en in paare:
    print(de, "/", en)
    for name, tokenizer in [("Qwen2.5", qwen), ("GPT-2", gpt2)]:
        n_de, n_en = len(tokenizer.encode(de)), len(tokenizer.encode(en))
        print(f"  {name:8} Deutsch {n_de:2} Token ({len(de) / n_de:.1f} Zeichen pro Token), "
              f"Englisch {n_en:2} ({len(en) / n_en:.1f}), Faktor {n_de / n_en:.1f}")
```

> **Aufgabe 4** (Bonus): Mit zwei Dingen haben Sprachmodelle oft Mühe: exakt rechnen und Buchstaben zählen. Nehmt zuerst eine lange Zahl, etwa `Long.MAX_VALUE`, einen Zeitstempel in Millisekunden oder einen Betrag: Wie zerlegen die beiden Tokenizer sie? Danach „strawberry“ in der Zelle weiter unten. Was sieht das Modell jeweils, und was heisst das, wenn es damit rechnen oder Buchstaben zählen soll?

```python
zahl = "Long.MAX_VALUE = 9223372036854775807"
zeige(zahl, qwen, gpt2)
```

```python
# @title Lösung (aufklappen)
# Qwen2.5 zerlegt jede Zahl in einzelne Ziffern, das legt eine Regel im Tokenizer fest.
# So machen es auch PaLM und Llama (Chowdhery et al. 2022, Touvron et al. 2023).
# GPT-2 zerlegt in Stücke, die in seinen Trainingsdaten häufig waren, nicht nach Stellen:
# 1000000 wird zu 1|000000, eins mehr zu 10000|01.
# Das Modell sieht also Token-IDs und keine Zahlen. Exakt rechnen fällt ihm schwer,
# ein Tool rechnet sicher.
for beispiel_zahl in ["1000000", "1000001"]:
    zeige(beispiel_zahl, qwen, gpt2)
print(1000000 + 1)
```

Wie viele r stecken in „strawberry“? Ihr seht es sofort, ein Modell bekommt nur die IDs. Zerlegt das Wort mit und ohne Leerzeichen davor, gross geschrieben und Buchstabe für Buchstabe. Warum hilft es, wenn ein Modell das Wort zuerst buchstabiert?

```python
for wort in [" strawberry", "strawberry", "Strawberry", "s t r a w b e r r y"]:
    zeige(wort, qwen)
```

```python
# @title Lösung (aufklappen)
# " strawberry" ist bei beiden Tokenizern ein einziges Token. Das Modell bekommt eine Zahl,
# nicht die zehn Buchstaben darin. Wie viele r in dieser ID stecken, steht nirgends in der
# Eingabe; das muss das Modell aus dem Training mitbringen.
# Ohne Leerzeichen oder gross geschrieben sind es andere IDs und andere Stücke.
# Buchstabiert ist jeder Buchstabe ein eigenes Token, die drei r haben dieselbe ID. Modelle,
# die vor der Antwort Zwischenschritte schreiben, zählen deshalb oft richtig: Sie schreiben
# das Wort zuerst Buchstabe für Buchstabe hin. Sicher ist ein Tool.
print(qwen.encode(" strawberry"), "gegen", qwen.encode("s t r a w b e r r y"))
print("strawberry".count("r"))
```

## Bonus: Knobelaufgabe zum Vokabular

Qwen2.5 kennt dreimal so viele Token wie GPT-2 und braucht oben für Deutsch und eingerückten Code weniger. Bei den Zahlen in Aufgabe 4 ist es umgekehrt, weil Qwen2.5 jede Ziffer einzeln nimmt.

> **Aufgabe 5** (Bonus, Knobelaufgabe): Findet einen Text **ohne Ziffern**, den GPT-2 in weniger Token zerlegt als Qwen2.5. Die Zelle prüft es selbst. Warum gibt es solche Texte, obwohl Qwen2.5 das grössere Vokabular hat?

```python
import unicodedata

text = "The build is green."  # euer Text, ohne Ziffern

zeige(text, qwen, gpt2)
n_qwen, n_gpt2 = len(qwen.encode(text)), len(gpt2.encode(text))
if any(unicodedata.category(zeichen).startswith("N") for zeichen in text):
    print("Der Text enthält Ziffern oder andere Zahlzeichen wie ½. Gesucht ist einer ohne.")
elif n_gpt2 < n_qwen:
    print(f"✓ GPT-2 braucht {n_gpt2} Token, Qwen2.5 {n_qwen}.")
else:
    print(f"Noch nicht: GPT-2 braucht {n_gpt2} Token, Qwen2.5 {n_qwen}.")
```

<details><summary>Tipp 1</summary>

Ein Vokabular wird aus Text gelernt: Was dort häufig vorkam, bekommt ein eigenes Token. GPT-2 ist auf englischen Webseiten trainiert, Qwen2.5 auf vielen Sprachen und viel Code. Welche Wörter hatten bei GPT-2 also eher Platz?

</details>
<details><summary>Tipp 2</summary>

Seltenere lange englische Wörter oder Namen, mit Leerzeichen davor, wie mitten im Satz. ` Massachusetts` ist bei beiden ein Token, ` unilaterally` nur bei GPT-2.

</details>

```python
# @title Lösung (aufklappen)
# Ein grösseres Vokabular enthält nicht einfach das kleinere. Welche Wörter ein eigenes Token
# bekommen, hängt vom Text ab, aus dem das Vokabular gelernt wurde. GPT-2 ist auf englischen
# Webseiten trainiert, nicht englische Seiten wurden entfernt (Radford et al. 2019). Qwen2.5
# verwendet den Tokenizer von Qwen, dessen Vokabular für Chinesisch und andere Sprachen
# erweitert wurde (Bai et al. 2023), und es enthält viel Code. Seltenere englische Wörter
# und Namen kennt deshalb oft nur GPT-2 als ein Token:
for wort in [" incomprehensible", " Schwarzenegger", " chimpanzees"]:
    zeige(wort, qwen, gpt2)
satz_loesung = "The incomprehensible chimpanzees unilaterally lobbied Schwarzenegger."
print(f"Ganzer Satz: Qwen2.5 {len(qwen.encode(satz_loesung))} Token, GPT-2 {len(gpt2.encode(satz_loesung))}\n")

# Die beiden Vokabulare im Vergleich: Token nur aus Buchstaben a bis z, mit oder ohne
# Leerzeichen davor, und Token mit chinesischen, japanischen oder koreanischen Zeichen.
import random
import re

nur_buchstaben = re.compile(r" ?[A-Za-z]+")
cjk = re.compile(r"[\u3040-\u30ff\u3400-\u4dbf\u4e00-\u9fff\uac00-\ud7af]")
stuecke = {name: t.batch_decode([[i] for i in range(len(t))]) for name, t in [("Qwen2.5", qwen), ("GPT-2", gpt2)]}
woerter = {name: {s for s in liste if nur_buchstaben.fullmatch(s)} for name, liste in stuecke.items()}
random.seed(0)
for name, anderer in [("GPT-2", "Qwen2.5"), ("Qwen2.5", "GPT-2")]:
    nur_hier = sorted(woerter[name] - woerter[anderer])
    beispiele = random.sample([w for w in nur_hier if w.startswith(" ") and len(w) > 8], 6)
    print(f"Nur {name} kennt {len(nur_hier):_} Token aus Buchstaben, z. B. {beispiele}".replace("_", "'"))
mit_cjk = {name: sum(1 for s in liste if cjk.search(s)) for name, liste in stuecke.items()}
print(f"Token mit chinesischen, japanischen oder koreanischen Zeichen: Qwen2.5 {mit_cjk['Qwen2.5']:_} von {len(qwen):_}, "
      f"GPT-2 {mit_cjk['GPT-2']:_} von {len(gpt2):_}".replace("_", "'"))
```

## Was ihr gesehen habt

Ein Sprachmodell sieht IDs aus einem festen Vokabular: Was beim Bau des Vokabulars häufig war, ist ein Token, etwa englische Wörter und Java-Schlüsselwörter, anderes wird zerlegt. Derselbe Inhalt auf Deutsch füllt deshalb das Kontextfenster schneller und kostet mehr. Auch wie das Modell Zahlen und einzelne Buchstaben sieht, legt der Tokenizer fest; darauf kommen wir bei den Grenzen eines LLM zurück.
