### `for` und `range`

```java
for (int i = 0; i < x.length; i++) { ... x[i] ... }

```

Python kennt nur die zweite Form. Wer den Index braucht, nimmt `range`. `zip` läuft über zwei Listen gleichzeitig und liefert Paare. `summe += 1` geht wie in Java, `++` gibt es nicht.

````python
x = [1, 3, 2]
w = [2, -1, 0.5]

for i in range(len(x)):   # range(3) ergibt 0, 1, 2
    print(i, x[i])

for xi, wi in zip(x, w):  # Paare (1, 2), (3, -1), (2, 0.5)
    print(xi, "mal", wi)
````

## Ein Neuron nachrechnen

Ein Neuron multipliziert jede Eingabe `x` mit ihrem Gewicht `w`, zählt alles zusammen und addiert den Bias `b`. Ist das Ergebnis grösser als null, gibt es 1 aus, sonst 0. In Java, wie auf der Folie zum Neuron:

```java
boolean perzeptron(double[] x) {
    return dot(w, x) + b > 0;
}
```

`dot(w, x)` ist das Skalarprodukt: jedes Gewicht mal seine Eingabe, alles zusammengezählt. Die Werte stammen von der Folie „Ein Neuron multipliziert, addiert und vergleicht“:

| | 1 | 2 | 3 |
|---|:-:|:-:|:-:|
| Eingabe `x` | 1 | 3 | 2 |
| Gewicht `w` | 2 | −1 | 0.5 |

Bias `b` = 0.5

Mit Bias muss eine 1 herauskommen. Ohne Bias (Defaultwert `b=0`) kommt eine 0 heraus: Die Summe ist dann genau null, und null ist nicht grösser als null.

````python
# Prüfungen für Aufgabe 2 und 3. Ausführen genügt, lesen müsst ihr sie nicht.

def pruefe_neuron(funktion):
    faelle = [  # Eingaben, Gewichte, Bias (None heisst ohne), erwartete Ausgabe, Hinweis
        ([1, 3, 2], [2, -1, 0.5], 0.5, 1, "Werte von der Folie"),
        ([1, 3, 2], [2, -1, 0.5], None, 0, "ohne Bias ist die Summe genau null"),
        ([1, 0], [1, 1], -0.5, 1, "Summe 0.5"),
        ([0, 0], [1, 1], -0.5, 0, "Summe -0.5"),
    ]
    richtig = 0
    for eingaben, gewichte, bias, erwartet, hinweis in faelle:
        if bias is None:
            aufruf, ergebnis = f"neuron({eingaben}, {gewichte})", funktion(eingaben, gewichte)
        else:
            aufruf, ergebnis = f"neuron({eingaben}, {gewichte}, {bias})", funktion(eingaben, gewichte, bias)
        if ergebnis == erwartet:
            richtig += 1
        else:
            print(f"✗ {aufruf} gibt {ergebnis}, erwartet {erwartet} ({hinweis})")
    zeichen = "✓ " if richtig == len(faelle) else ""
    print(f"{zeichen}{richtig} von {len(faelle)} Fällen richtig")


def pruefe_gewicht(w_neu):
    w_alt, x_alt, b_alt = [2, -1, 0.5], [1, 3, 2], 0.5
    summe = round(x_alt[0] * w_neu[0] + x_alt[1] * w_neu[1] + x_alt[2] * w_neu[2] + b_alt, 9)
    ausgabe = 1 if summe > 0 else 0
    geaendert = [f"w{i + 1}: {w_alt[i]} → {w_neu[i]}" for i in range(3) if w_neu[i] != w_alt[i]]
    zeile = f"Summe {summe:g}, Ausgabe {ausgabe}"
    if not geaendert:
        print(f"{zeile}. Noch kein Gewicht geändert.")
    elif len(geaendert) > 1:
        print(f"{zeile}. Geändert: {', '.join(geaendert)}. Ändert genau eines.")
    elif ausgabe == 0:
        print(f"✓ {geaendert[0]}, {zeile}")
    else:
        print(f"{geaendert[0]}, {zeile}. Die Summe muss auf null oder darunter.")
````

> **Aufgabe 2**: Ergänzt `neuron` in der nächsten Zelle, sodass die Funktion die Summe ausrechnet und 0 oder 1 zurückgibt. Die letzte Zeile prüft eure Funktion mit vier Fällen, darunter die zwei von der Folie. Fertig, wenn darunter „✓ 4 von 4 Fällen richtig“ steht.

````python
x = [1, 3, 2]      # Eingaben
w = [2, -1, 0.5]   # Gewichte
b = 0.5            # Bias


def neuron(x, w, b=0):
    summe = 0
    # hier: jede Eingabe mal ihr Gewicht aufsummieren, dann den Bias dazu
    return 1 if summe > 0 else 0


pruefe_neuron(neuron)
````

<details><summary>Tipp 1</summary>

Über zwei Listen gleichzeitig läuft `zip`, wie im Abschnitt zu `for` und `range`. In der Schleife wächst `summe` um Eingabe mal Gewicht, nach der Schleife kommt der Bias dazu.

</details>
<details><summary>Tipp 2</summary>

```python
    for xi, wi in zip(x, w):
        summe += ...
    summe += ...
```

</details>

````python
# @title Lösung (aufklappen)
def neuron_loesung(x, w, b=0):
    summe = 0
    for xi, wi in zip(x, w):
        summe += xi * wi
    summe += b
    return 1 if summe > 0 else 0


# Werte von der Folie, eigene Namen, damit eure Variablen unberührt bleiben
x_folie = [1, 3, 2]
w_folie = [2, -1, 0.5]
b_folie = 0.5

# Dieselbe Rechnung Schritt für Schritt wie auf der Folie
dot = 0
for xi, wi in zip(x_folie, w_folie):
    print(f"{xi} × {wi} = {xi * wi}")
    dot += xi * wi
print(f"Summe dot(w, x) = {dot}, mit Bias {dot + b_folie}")
print(f"Ausgabe mit Bias: {neuron_loesung(x_folie, w_folie, b_folie)}")
print(f"Ausgabe ohne Bias: {neuron_loesung(x_folie, w_folie)}")

pruefe_neuron(neuron_loesung)
````

## Ein Gewicht ändern

Die Summe mit Bias ist 0.5. Damit das Neuron eine 0 ausgibt, muss sie auf null oder darunter. Die Zelle rechnet die Summe selbst aus, eure Funktion `neuron` muss dafür nicht fertig sein.

> **Aufgabe 3**: Ändert in `w_neu` genau ein Gewicht so, dass das Neuron mit Bias 0.5 eine 0 ausgibt. Probiert es für jedes der drei Gewichte. Welches kommt mit der kleinsten Änderung aus, und warum? Fertig, wenn ihr für jedes Gewicht ein ✓ hattet und die Frage beantworten könnt.

````python
w_neu = [2, -1, 0.5]   # hier genau ein Gewicht ändern
pruefe_gewicht(w_neu)
````

<details><summary>Tipp</summary>

Macht jedes Gewicht um 0.1 kleiner und vergleicht, wie stark die Summe jeweils sinkt. Womit wird das Gewicht multipliziert?

</details>

````python
# @title Lösung (aufklappen)
# Ein Gewicht verschiebt die Summe um (Änderung × seine Eingabe).
# Die Summe muss um 0.5 sinken, also muss Gewicht i um mindestens 0.5 / x[i] kleiner werden.
# Genau auf der Grenze ist die Summe null, und das gibt auch schon eine 0.
# Das zweite Gewicht hängt an der grössten Eingabe (3), deshalb reicht dort die kleinste Änderung.
# Diese Grösse braucht später das Training: Wie stark die Summe auf ein Gewicht reagiert,
# ist die Eingabe, die an diesem Gewicht hängt.

for i in range(len(w_folie)):
    noetig = 0.5 / x_folie[i]
    w_probe = list(w_folie)   # Kopie, wie new ArrayList<>(w)
    w_probe[i] = round(w_folie[i] - noetig - 0.01, 2)   # 0.01 mehr als nötig
    print(f"w{i + 1} an Eingabe {x_folie[i]}: mindestens {noetig:.2f} kleiner, "
          f"{w_folie[i]} → {w_probe[i]} gibt Ausgabe {neuron_loesung(x_folie, w_probe, b_folie)}")
````

## Bonus: das XOR-Netz

XOR ist in Java `a ^ b`: 1, wenn genau eine der beiden Eingaben 1 ist. Ein einzelnes Neuron schafft das nicht, es zieht nur eine Gerade, und die beiden Einsen liegen diagonal gegenüber. Mit zwei Neuronen in der Hidden Layer geht es, wie auf der Folie „XOR braucht eine zweite Schicht“:

- `h₁` meldet 1, wenn mindestens eine Eingabe 1 ist (ODER)
- `h₂` meldet 1, wenn beide Eingaben 1 sind (UND)
- der Ausgang `y` meldet 1, wenn `h₁` ja sagt und `h₂` nein

Die Folie zeigt die Grenzen von `h₁` und `h₂` als Geraden, ohne Zahlen. Sie liegen bei x₁ + x₂ = 0.5 und x₁ + x₂ = 1.5. Daraus folgen diese Gewichte:

| Neuron | Eingaben | Gewichte `w` | Bias `b` |
|---|---|:-:|:-:|
| `h₁` (ODER) | x₁, x₂ | 1, 1 | −0.5 |
| `h₂` (UND) | x₁, x₂ | 1, 1 | −1.5 |
| `y` | h₁, h₂ | 1, −1 | −0.5 |

Als Aktivierung dient überall die Stufe aus `neuron`: grösser als null gibt 1. Die Folie nennt ReLU als heute übliche Aktivierung; was mit ReLU passiert, ist Aufgabe 5.

> **Aufgabe 4** (Bonus): Rechnet das Netz für alle vier Eingaben nach, zuerst mit eurer Funktion `neuron` in einer Schleife, dann mit NumPy in einer Zeile. Fertig, wenn beide Zellen ✓ bei allen vier Eingaben zeigen.

Falls `neuron` die Prüfung oben noch nicht besteht: zuerst `neuron = neuron_loesung` ausführen.

````python
eingaben = [[0, 0], [0, 1], [1, 0], [1, 1]]

for x1, x2 in eingaben:   # jede Zeile in zwei Variablen zerlegt
    h1 = neuron([x1, x2], [1, 1], -0.5)
    h2 = 0   # hier: das UND-Neuron
    y = 0    # hier: der Ausgang aus h1 und h2
    zeichen = "✓" if y == x1 ^ x2 else "✗"
    print(f"{zeichen} {x1} ^ {x2}: Netz {y}, erwartet {x1 ^ x2}   (h₁ = {h1}, h₂ = {h2})")
````

<details><summary>Tipp</summary>

`h2` geht wie `h1`, nur mit dem Bias des UND-Neurons aus der Tabelle. Der Ausgang ist ein drittes Neuron, seine Eingaben sind `[h1, h2]`.

</details>

````python
# @title Lösung (aufklappen)
for x1, x2 in [[0, 0], [0, 1], [1, 0], [1, 1]]:
    h1_l = neuron_loesung([x1, x2], [1, 1], -0.5)
    h2_l = neuron_loesung([x1, x2], [1, 1], -1.5)
    y_l = neuron_loesung([h1_l, h2_l], [1, -1], -0.5)
    zeichen_l = "✓" if y_l == x1 ^ x2 else "✗"
    print(f"{zeichen_l} {x1} ^ {x2}: Netz {y_l}, erwartet {x1 ^ x2}   (h₁ = {h1_l}, h₂ = {h2_l})")
````

**Zweiter Teil von Aufgabe 4:** Dasselbe mit NumPy, für alle vier Eingaben auf einmal.

In NumPy wird eine Schicht zu einer Matrixmultiplikation. `X` hält die vier Eingaben als Zeilen, jede Zeile von `W1` die Gewichte eines Neurons. `X @ W1.T` rechnet alle Skalarprodukte auf einmal: vier Eingaben mal zwei Neuronen, eine 4 × 2-Matrix. `+ b1` addiert den Bias zu jeder Zeile (das heisst Broadcasting). In Java wären das zwei verschachtelte `for`-Schleifen. PyTorch speichert die Gewichte einer Schicht (`nn.Linear`) genauso, eine Zeile pro Neuron, und rechnet `x @ W.T + b`.

Die Zelle rechnet die Hidden Layer schon aus. Ergänzt `y` in einer Zeile; die letzte Zeile vergleicht es mit XOR.

````python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])   # vier Eingaben, eine pro Zeile
W1 = np.array([[1, 1],                          # h₁
               [1, 1]])                         # h₂
b1 = np.array([-0.5, -1.5])
W2 = np.array([1, -1])                          # Ausgang y
b2 = -0.5


def stufe_np(z):
    return (z > 0).astype(int)   # für alle Einträge auf einmal


H = stufe_np(X @ W1.T + b1)   # eine Zeile pro Eingabe, eine Spalte pro Neuron
print(H)

y = None   # hier: y in einer Zeile

erwartet = X[:, 0] ^ X[:, 1]   # XOR der ersten und der zweiten Spalte
print(y, "✓" if np.array_equal(y, erwartet) else f"erwartet {erwartet}")
````

<details><summary>Tipp</summary>

Der Ausgang ist eine Schicht mit einem Neuron. Er rechnet wie die Zeile für `H`, nur mit `H`, `W2` und `b2` statt `X`, `W1` und `b1`. `W2` ist eine einzelne Zeile, deshalb braucht es kein `.T`.

</details>

````python
# @title Lösung (aufklappen)
y_loesung = stufe_np(stufe_np(X @ W1.T + b1) @ W2 + b2)
print(y_loesung)
print(X[:, 0] ^ X[:, 1])   # zum Vergleich: XOR der ersten und der zweiten Spalte
````

## Bonus: XOR mit ReLU

ReLU gibt bis null eine 0 aus und darüber den Wert unverändert weiter, statt ihn auf 1 zu setzen: `relu(z) = max(0, z)`, in Java `Math.max(0, z)`. Das ist die Aktivierung, die man heute meistens nimmt.

> **Aufgabe 5** (Bonus, Knobelaufgabe): Die nächste Zelle setzt ReLU in die Hidden Layer, mit den Gewichten von oben; der Ausgang bleibt eine Stufe. Erst vermuten: Rechnet das Netz damit noch XOR? Dann ausführen und für jede Zeile mit ✗ herausfinden, warum. Ändert danach Gewichte und Bias in der Zelle, bis alle vier Zeilen ✓ zeigen. ReLU bleibt in der Hidden Layer.

Die Zelle braucht `X` und `stufe_np` aus der NumPy-Zelle von Aufgabe 4.

````python
def relu(z):
    return np.maximum(0, z)   # wie Math.max(0, z), für alle Einträge auf einmal


# Gewichte von oben, hier ändern
W1_relu = np.array([[1, 1],    # h₁
                    [1, 1]])   # h₂
b1_relu = np.array([-0.5, -1.5])
W2_relu = np.array([1, -1])    # Ausgang y
b2_relu = -0.5

H_relu = relu(X @ W1_relu.T + b1_relu)         # Hidden Layer mit ReLU
y_relu = stufe_np(H_relu @ W2_relu + b2_relu)  # Ausgang weiter mit Stufe

for (x1, x2), (h1, h2), yi in zip(X, H_relu, y_relu):
    zeichen = "✓" if yi == x1 ^ x2 else "✗"
    print(f"{zeichen} {x1} ^ {x2}: Netz {yi}, erwartet {x1 ^ x2}   (h₁ = {h1:g}, h₂ = {h2:g})")
````

<details><summary>Tipp 1</summary>

Vergleicht h₁ und h₂ mit Aufgabe 4. Mit der Stufe waren sie 0 oder 1, mit ReLU kommen 0.5 und 1.5 heraus. Was muss der Ausgang aus diesen Werten machen, damit (0, 1) eine 1 ergibt und (1, 1) eine 0?

</details>
<details><summary>Tipp 2</summary>

Zwei Wege. Entweder ihr ändert `b1_relu` so, dass h₁ und h₂ wieder ganze Zahlen melden: h₁ = relu(x₁ + x₂), h₂ = relu(x₁ + x₂ − 1). Oder ihr lasst die Hidden Layer, wie sie ist, und passt nur `W2_relu` und `b2_relu` an. In beiden Fällen muss h₂ bei (1, 1) das h₁ ausgleichen, braucht also ein stärker negatives Gewicht als −1.

</details>

````python
# @title Lösung (aufklappen)
# Mit der Stufe melden h₁ und h₂ nur 0 oder 1. ReLU gibt weiter, wie weit die Summe über null
# liegt: bei einer Eins h₁ = 0.5, bei zwei Einsen h₁ = 1.5 und h₂ = 0.5. Der Ausgang war auf 0
# und 1 eingestellt. Bei (0, 1) landet er mit 0.5 − 0.5 genau auf null und gibt 0, bei (1, 1)
# gleicht h₂ = 0.5 das h₁ = 1.5 nicht mehr aus. Drei von vier Zeilen kippen.
#
# Eine Lösung: h₁ = relu(x₁ + x₂) zählt die Einsen, h₂ = relu(x₁ + x₂ − 1) meldet 1, wenn es
# zwei sind, und y = stufe(h₁ − 2·h₂ − 0.5) zieht in dem Fall beide wieder ab. Eine andere lässt
# die Hidden Layer, wie sie ist: y = stufe(h₁ − 3·h₂ − 0.25). Passende Gewichte gibt es viele.
#
# Warum man trotzdem ReLU nimmt: Mit der Stufe ändert ein kleiner Dreh an einem Gewicht den
# Ausgang meistens gar nicht, und das Training findet keine Richtung. Mit ReLU ändert sich die
# Ausgabe eines Neurons mit, solange seine Summe über null liegt.

erwartet_loesung = X[:, 0] ^ X[:, 1]
varianten_loesung = {
    "Bias der Hidden Layer verschoben": (np.array([[1, 1], [1, 1]]), np.array([0, -1]), np.array([1, -2]), -0.5),
    "nur der Ausgang angepasst": (np.array([[1, 1], [1, 1]]), np.array([-0.5, -1.5]), np.array([1, -3]), -0.25),
}
for titel_l, (W1_l, b1_l, W2_l, b2_l) in varianten_loesung.items():
    H_l = relu(X @ W1_l.T + b1_l)
    y_l = stufe_np(H_l @ W2_l + b2_l)
    zeichen_l = "✓" if np.array_equal(y_l, erwartet_loesung) else "✗"
    print(f"{zeichen_l} {titel_l}: Netz {y_l}, h₁ = {H_l[:, 0]}, h₂ = {H_l[:, 1]}")
````

## Was ihr gesehen habt

Ein Neuron multipliziert jede Eingabe mit ihrem Gewicht, zählt alles zusammen, addiert den Bias und vergleicht mit null. Wie stark ein Gewicht die Summe bewegt, hängt an der Eingabe, mit der es multipliziert wird. Das XOR-Netz besteht aus drei solchen Neuronen mit neun Parametern, sechs Gewichten und drei Bias-Werten. Das MNIST-Netz im Workshop ist genauso gebaut, nur mit 784 Eingängen und 32 Neuronen in der Mitte, und dort setzt das Training die Gewichte, nicht ihr von Hand.
