## 1. Python in sechs Zellen

### Variablen ohne Typ

```java
int anzahl = 3;
double bias = 0.5;
String name = "Neuron";
boolean feuert = true;
```

In Python hängt der Typ am Wert, nicht an der Variable. Kein Semikolon, `True` und `False` gross, `None` statt `null`, Kommentare mit `#`. Typangaben wie `bias: float = 0.5` sind möglich, Python prüft sie beim Ausführen aber nicht.

````python
anzahl = 3
bias = 0.5
name = "Neuron"
feuert = True
print(type(anzahl), type(bias), type(name), type(feuert))

anzahl = "drei"   # erlaubt: die Variable zeigt jetzt auf einen String
print(type(anzahl))
````

### Einrückung statt Klammern

```java
if (summe > 0) {
    ausgabe = 1;
} else {
    ausgabe = 0;
}
```

Der Doppelpunkt öffnet den Block, die Einrückung mit vier Leerzeichen ist der Block. Logisch verknüpft wird mit `and`, `or` und `not` statt `&&`, `||` und `!`.

````python
summe = 0.5

if summe > 0:
    ausgabe = 1
else:
    ausgabe = 0
print(ausgabe)

ausgabe = 1 if summe > 0 else 0   # wie summe > 0 ? 1 : 0
print(ausgabe)
````

### Listen und Slices

```java
double[] x = {1, 3, 2};
x[0];  x.length;
List.of(1, 3, 2).subList(0, 2);
```

Eine Python-Liste wächst wie eine `ArrayList`. Der Index beginnt bei 0, ein negativer Index zählt von hinten. Ein Slice `[von:bis]` schneidet einen Teil heraus, `bis` ist exklusiv wie bei `subList`.

````python
x = [1, 3, 2]
print(x[0], x[-1], len(x))   # erstes, letztes, Länge
print(x[0:2], x[1:])         # von inklusive, bis exklusiv; ohne bis: bis zum Ende
x.append(4)                  # wie list.add(4)
print(x)
````

### `for` und `range`

```java
for (int i = 0; i < x.length; i++) { ... x[i] ... }
for (double xi : x) { ... }
```

Python kennt nur die zweite Form. Wer den Index braucht, nimmt `range`. `zip` läuft über zwei Listen gleichzeitig und liefert Paare. `summe += 1` geht wie in Java, `++` gibt es nicht.

````python
x = [1, 3, 2]
w = [2, -1, 0.5]

for i in range(len(x)):   # range(3) ergibt 0, 1, 2
    print(i, x[i])

for xi, wi in zip(x, w):  # Paare (1, 2), (3, -1), (2, 0.5)
    print(xi, "mal", wi)
````

### `def` mit Defaultwerten

```java
int stufe(double summe) { return stufe(summe, 0); }
int stufe(double summe, double schwelle) { return summe > schwelle ? 1 : 0; }
```

Statt zu überladen, gibt man einem Parameter einen Defaultwert. Beim Aufruf darf man Parameter beim Namen nennen. Typen stehen keine da, der Rückgabetyp auch nicht.

````python
def stufe(summe, schwelle=0):
    return 1 if summe > schwelle else 0


print(stufe(0.5))               # Schwelle 0
print(stufe(0.5, schwelle=1))   # Parameter beim Namen
````

### f-Strings

```java
String.format("Summe %.2f, Ausgabe %d", summe, ausgabe)
```

Ein `f` vor dem String, Ausdrücke in geschweiften Klammern, das Format nach dem Doppelpunkt.

````python
summe = 0.5
print(f"Summe {summe:.2f}, Ausgabe {stufe(summe)}")
print(f"{summe = }")   # Name und Wert, praktisch zum Debuggen
````

## 2. Pflicht: ein Neuron nachrechnen

Ein Neuron multipliziert jede Eingabe `x` mit ihrem Gewicht `w`, zählt alles zusammen und addiert den Bias `b`. Ist das Ergebnis grösser als null, gibt es 1 aus, sonst 0. In Java, wie auf der Folie zum Neuron:

```java
boolean perzeptron(double[] x) {
    return dot(w, x) + b > 0;
}
```

`dot(w, x)` ist das Skalarprodukt: jedes Gewicht mal seine Eingabe, alles zusammengezählt. Die Werte stammen von der Folie „Ein Neuron multipliziert, addiert und vergleicht“:

| | 1 | 2 | 3 |
|---|:-:|:-:|:-:|
| Eingabe `x` | 1 | 3 | 2 |
| Gewicht `w` | 2 | −1 | 0.5 |

Bias `b` = 0.5

**Aufgabe:** Ergänzt `neuron`, sodass die Funktion die Summe ausrechnet und 0 oder 1 zurückgibt. Mit Bias muss eine 1 herauskommen. Ohne Bias (Defaultwert `b=0`) kommt eine 0 heraus: Die Summe ist dann genau null, und null ist nicht grösser als null.

````python
x = [1, 3, 2]      # Eingaben
w = [2, -1, 0.5]   # Gewichte
b = 0.5            # Bias


def neuron(x, w, b=0):
    summe = 0
    # hier: jede Eingabe mal ihr Gewicht aufsummieren, dann den Bias dazu
    return 1 if summe > 0 else 0


print(neuron(x, w, b))   # erwartet: 1
print(neuron(x, w))      # erwartet: 0
````

````python
# @title Lösung (aufklappen)
def neuron_loesung(x, w, b=0):
    summe = 0
    for xi, wi in zip(x, w):
        summe += xi * wi
    summe += b
    return 1 if summe > 0 else 0


# Werte von der Folie, eigene Namen, damit eure Variablen unberührt bleiben
x_folie = [1, 3, 2]
w_folie = [2, -1, 0.5]
b_folie = 0.5

# Dieselbe Rechnung Schritt für Schritt wie auf der Folie
dot = 0
for xi, wi in zip(x_folie, w_folie):
    print(f"{xi} × {wi} = {xi * wi}")
    dot += xi * wi
print(f"Summe dot(w, x) = {dot}, mit Bias {dot + b_folie}")
print(f"Ausgabe mit Bias: {neuron_loesung(x_folie, w_folie, b_folie)}")
print(f"Ausgabe ohne Bias: {neuron_loesung(x_folie, w_folie)}")
````

## 3. Pflicht: ein Gewicht ändern

**Aufgabe:** Ändert genau ein Gewicht so, dass das Neuron mit Bias 0.5 eine 0 ausgibt. Probiert es für jedes der drei Gewichte. Welches kommt mit der kleinsten Änderung aus, und warum?

Die Summe mit Bias ist 0.5, sie muss auf null oder darunter. Falls eure Funktion `neuron` noch nicht stimmt, nehmt `neuron_loesung` aus der Lösung oben.

````python
w_neu = [2, -1, 0.5]   # hier ein Gewicht ändern
print(neuron(x, w_neu, b))
````

````python
# @title Lösung (aufklappen)
# Ein Gewicht verschiebt die Summe um (Änderung × seine Eingabe).
# Die Summe muss um 0.5 sinken, also muss Gewicht i um mindestens 0.5 / x[i] kleiner werden.
# Genau auf der Grenze ist die Summe null, und das gibt auch schon eine 0.
# Das zweite Gewicht hängt an der grössten Eingabe (3), deshalb reicht dort die kleinste Änderung.
# Diese Grösse braucht später das Training: Wie stark die Summe auf ein Gewicht reagiert,
# ist die Eingabe, die an diesem Gewicht hängt.

for i in range(len(w_folie)):
    noetig = 0.5 / x_folie[i]
    w_probe = list(w_folie)   # Kopie, wie new ArrayList<>(w)
    w_probe[i] = round(w_folie[i] - noetig - 0.01, 2)   # 0.01 mehr als nötig
    print(f"w{i + 1} an Eingabe {x_folie[i]}: mindestens {noetig:.2f} kleiner, "
          f"{w_folie[i]} → {w_probe[i]} gibt Ausgabe {neuron_loesung(x_folie, w_probe, b_folie)}")
````

## 4. Zusatz: das XOR-Netz

XOR ist in Java `a ^ b`: 1, wenn genau eine der beiden Eingaben 1 ist. Ein einzelnes Neuron schafft das nicht, es zieht nur eine Gerade, und die beiden Einsen liegen diagonal gegenüber. Mit zwei Neuronen in der Hidden Layer geht es, wie auf der Folie „XOR braucht eine zweite Schicht“:

- `h₁` meldet 1, wenn mindestens eine Eingabe 1 ist (ODER)
- `h₂` meldet 1, wenn beide Eingaben 1 sind (UND)
- der Ausgang `y` meldet 1, wenn `h₁` ja sagt und `h₂` nein

Die Folie zeigt die Grenzen von `h₁` und `h₂` als Geraden, ohne Zahlen. Sie liegen bei x₁ + x₂ = 0.5 und x₁ + x₂ = 1.5. Daraus folgen diese Gewichte:

| Neuron | Eingaben | Gewichte `w` | Bias `b` |
|---|---|:-:|:-:|
| `h₁` (ODER) | x₁, x₂ | 1, 1 | −0.5 |
| `h₂` (UND) | x₁, x₂ | 1, 1 | −1.5 |
| `y` | h₁, h₂ | 1, −1 | −0.5 |

Als Aktivierung dient überall die Stufe aus `neuron`: grösser als null gibt 1. Die Folie nennt ReLU als heute übliche Aktivierung. Hier bleibt es bei der Stufe, damit `h₁` und `h₂` genau 0 oder 1 melden; mit ReLU bräuchte das Netz andere Gewichte.

**Aufgabe a:** Rechnet das Netz mit eurer Funktion `neuron` für alle vier Eingaben nach. Falls sie noch nicht stimmt: zuerst `neuron = neuron_loesung` ausführen.

````python
eingaben = [[0, 0], [0, 1], [1, 0], [1, 1]]

for x1, x2 in eingaben:   # jede Zeile in zwei Variablen zerlegt
    h1 = neuron([x1, x2], [1, 1], -0.5)
    h2 = 0   # hier: das UND-Neuron
    y = 0    # hier: der Ausgang aus h1 und h2
    print(f"{x1} ^ {x2}: Netz {y}, erwartet {x1 ^ x2}   (h₁ = {h1}, h₂ = {h2})")
````

````python
# @title Lösung (aufklappen)
for x1, x2 in [[0, 0], [0, 1], [1, 0], [1, 1]]:
    h1_l = neuron_loesung([x1, x2], [1, 1], -0.5)
    h2_l = neuron_loesung([x1, x2], [1, 1], -1.5)
    y_l = neuron_loesung([h1_l, h2_l], [1, -1], -0.5)
    print(f"{x1} ^ {x2}: Netz {y_l}, erwartet {x1 ^ x2}   (h₁ = {h1_l}, h₂ = {h2_l})")
````

**Aufgabe b:** Dasselbe mit NumPy, für alle vier Eingaben auf einmal.

In NumPy wird eine Schicht zu einer Matrixmultiplikation. `X` hält die vier Eingaben als Zeilen, jede Zeile von `W1` die Gewichte eines Neurons. `X @ W1.T` rechnet alle Skalarprodukte auf einmal: vier Eingaben mal zwei Neuronen, eine 4 × 2-Matrix. `+ b1` addiert den Bias zu jeder Zeile (das heisst Broadcasting). In Java wären das zwei verschachtelte `for`-Schleifen. PyTorch speichert die Gewichte einer Schicht (`nn.Linear`) genauso, eine Zeile pro Neuron, und rechnet `x @ W.T + b`.

Die Zelle rechnet die Hidden Layer schon aus. Schreibt darunter `y` in einer Zeile, aus `X`, `W1`, `b1`, `W2` und `b2`.

````python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])   # vier Eingaben, eine pro Zeile
W1 = np.array([[1, 1],                          # h₁
               [1, 1]])                         # h₂
b1 = np.array([-0.5, -1.5])
W2 = np.array([1, -1])                          # Ausgang y
b2 = -0.5


def stufe_np(z):
    return (z > 0).astype(int)   # für alle Einträge auf einmal


H = stufe_np(X @ W1.T + b1)   # eine Zeile pro Eingabe, eine Spalte pro Neuron
print(H)

# hier: y in einer Zeile
````

````python
# @title Lösung (aufklappen)
y_loesung = stufe_np(stufe_np(X @ W1.T + b1) @ W2 + b2)
print(y_loesung)
print(X[:, 0] ^ X[:, 1])   # zum Vergleich: XOR der ersten und der zweiten Spalte
````

Damit habt ihr ein vollständiges neuronales Netz gerechnet: neun Parameter, sechs Gewichte und drei Bias-Werte. Das MNIST-Netz im Workshop ist genauso gebaut, nur mit 784 Eingängen und 32 Neuronen in der Mitte. Dort setzt das Training die Gewichte, nicht ihr von Hand.
